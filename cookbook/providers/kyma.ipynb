## With OpenAI Client

```python
from openai import OpenAI
from portkey_ai import PORTKEY_GATEWAY_URL, createHeaders

client = OpenAI(
    base_url=PORTKEY_GATEWAY_URL,
    default_headers=createHeaders(
        provider="kyma",
        api_key="YOUR_KYMA_API_KEY"
    )
)

chat_completion = client.chat.completions.create(
    messages=[{"role": "user", "content": "What is the meaning of life?"}],
    model="deepseek-v3"  # Try: deepseek-r1, llama-3.3-70b, qwen-3-32b, gemma-4-31b
)

print(chat_completion.choices[0].message.content)
```

## With Portkey Client

```python
from portkey_ai import Portkey

portkey = Portkey(
    api_key="YOUR_PORTKEY_API_KEY",
    provider="kyma",
    Authorization="YOUR_KYMA_API_KEY"
)

completion = portkey.chat.completions.create(
    messages=[{"role": "user", "content": "Explain quantum computing in simple terms"}],
    model="llama-3.3-70b"
)

print(completion.choices[0].message.content)
```

## Streaming Responses

```python
stream = portkey.chat.completions.create(
    messages=[{"role": "user", "content": "Write a short poem about open-source AI"}],
    model="deepseek-v3",
    stream=True
)

for chunk in stream:
    if chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="")
```

## Available Models

Kyma API serves 100+ models behind the same endpoint. A few commonly used chat model IDs:

- `qwen-3.6-plus`: strong general-purpose and coding model
- `deepseek-v3`: fast, cost-effective general model
- `deepseek-r1`: reasoning model
- `kimi-k2.6`: long-context agentic model with image input
- `llama-3.3-70b`: popular open-weight baseline
- `gemma-4-31b`: multimodal open model
- `qwen-3-coder`: code generation
- `gemini-3-flash`: 1M-token context window

Kyma also accepts aliases such as `best`, `fast`, `code` and `cheap`, which resolve to a current pick for that category.
